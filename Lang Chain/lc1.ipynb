```python
# from langchain_community.llms import ollama
from langchain_ollama import OllamaLLM
from langchain import PromptTemplate, LLMChain
from langchain.chains import SimpleSequentialChain, SequentialChain
```

```python
llm = OllamaLLM(model="llama3.2", temperature=0.9)
```

```python
prompt = PromptTemplate(
    input_variables=["product"],
    template="What is the best name to describe a company that makes {product}? just give name",
)
# llmNameChain = prompt | llm
llmNameChain = LLMChain(llm=llm, prompt=prompt, output_key="name")
# llmNameChain.invoke('pizza')
```

```python
promptDesc = PromptTemplate(
    input_variables=["name"],
    template="Write a 20 words description for the following company:{name}",
)
# llmDescChain = promptDesc | llm
llmDescChain = LLMChain(llm=llm, prompt=promptDesc, output_key="description")
# llmDescChain.invoke()
```

```python
llmChain = SimpleSequentialChain(chains=[llmNameChain, llmDescChain], )
response = llmChain.invoke("pizza")
```

```python
response
```

```python
llmChain = SequentialChain(
    chains=[llmNameChain, llmDescChain],
    input_variables=["product"],
    output_variables=["name", "description"],
    verbose=True,
)
llmChain.invoke("laptop")
```
